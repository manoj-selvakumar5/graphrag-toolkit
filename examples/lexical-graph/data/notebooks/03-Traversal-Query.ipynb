```python
%reload_ext dotenv
%dotenv ../.env


import os

from graphrag_toolkit.lexical_graph import set_logging_config
from graphrag_toolkit.lexical_graph import LexicalGraphQueryEngine
from graphrag_toolkit.lexical_graph.storage import GraphStoreFactory
from graphrag_toolkit.lexical_graph.storage import VectorStoreFactory
set_logging_config('INFO')

# Initialize stores and index
graph_store = GraphStoreFactory.for_graph_store(os.environ['GRAPH_STORE'])
vector_store = VectorStoreFactory.for_vector_store(os.environ['VECTOR_STORE'])



# Initialize query engine
query_engine = LexicalGraphQueryEngine.for_traversal_based_search(
    graph_store, 
    vector_store,
    streaming=True
)
```

```python

# Example 1: Company and Technology query
print("Example 1: Company and Technology query")
response = query_engine.query(
    "What are the major semiconductor companies and what technologies do they produce?",
)
print("\nResponse:", response)
```

```python
import json
for n in response.source_nodes:
    print(json.dumps(n.metadata, indent=2))
```

```python

# Example 2: Supply Chain Relationship query
print("\nExample 2: Supply Chain Relationship query")
response = query_engine.query(
    "How does Taiwan's semiconductor industry affect global supply chains?",
)
print("\nResponse:", response)
```

```python

# Example 3: Policy Impact query
print("\nExample 3: Policy Impact query")
response = query_engine.query(
    "What is the CHIPS Act and how does it regulate semiconductor manufacturing?",
)
print("\nResponse:", response)
```

```python
import json
for n in response.source_nodes:
    print(json.dumps(n.metadata, indent=2))
```

```python

# Example 4: Resource Dependencies query
print("\nExample 4: Resource Dependencies query")
response = query_engine.query(
    "What raw materials are critical for semiconductor manufacturing and which countries supply them?",
    show_intermediate_steps=True
)
print("\nResponse:", response)
```
